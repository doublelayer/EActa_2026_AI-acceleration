```python
#########################
##-Importing libraries-##
#########################

##-General
import re
from pathlib import Path as FilePath
import numpy as np

##-Matplotlib
import matplotlib.pyplot as plt

from matplotlib.path import Path
from matplotlib.patches import PathPatch
from matplotlib.colors import to_rgba
from matplotlib.transforms import Affine2D
```

```python
####################
##-Plotting style-##
####################

plt.rc('figure', figsize=(17 / 2.54, 7.18 / 2.54))
plt.rc('figure', dpi=600)
plt.rc('font', size=9, family='DejaVu Sans')
plt.rc('svg', fonttype='none')
plt.rc('pdf', fonttype=42)

green = '#50D050'
```

```python
#######################
##-Arrow and labels-##
#######################

svg_arrow = '''
m 345.11,280.35
c 15.53,15.53 40.71,15.53 56.23,0 15.53,-15.53 15.53,-40.71 0,-56.24 -5.35,-5.35 -12.12,-9.06 -19.51,-10.7
l -0.44,-6.22 -10.61,10.44 12.24,12.63 -0.43,-6.12
c 15.51,5.17 23.9,21.94 18.73,37.45 -5.17,15.52 -21.94,23.91 -37.45,18.74 -4.37,-1.46 -8.33,-3.91 -11.58,-7.16
z
'''

stages = ['Brainstorm', 'Structure', 'Draft', 'Revise', 'Polish', 'Publish', 'Share']

centers = [(172, 315), (311, 343), (416, 249), (554, 280), (661, 190), (795, 214), (904, 120)]

notes = [
    (16, 138, 'left', 'Research data,\nData analysis,\nScope and readership,\nFigures,\nIdeas'),
    (311, 456, 'center', 'Sections, plot, story'),
    (416, 100, 'center', 'Suggestions,\nComments'),
    (554, 390, 'center', 'Clarity,\nReadability,\nStructure'),
    (661, 30, 'center', 'Proofread,\nSanity check'),
    (795, 331, 'center', 'Journal Guidelines,\nSubmission,\nArchiving'),
    (1065, 30, 'center', 'Open Access,\nFAIR data,\nSocial Media'),
]


##-Convert the SVG coordinates into a Matplotlib path
tokens = re.findall(r'[mclz]|[-+]?(?:\d*\.\d+|\d+)', svg_arrow)

current = np.zeros(2)
vertices = []
codes = []
i = 0

while i < len(tokens):
    if tokens[i] in ['m', 'c', 'l', 'z']:
        command = tokens[i]
        i += 1

    if command == 'z':
        vertices.append(start.copy())
        codes.append(Path.CLOSEPOLY)
        break

    count = 6 if command == 'c' else 2
    values = np.array(tokens[i:i + count], dtype=float).reshape(-1, 2)
    points = current + values

    vertices.extend(points)

    if command == 'c':
        codes.extend([Path.CURVE4] * 3)
    elif command == 'm':
        codes.append(Path.MOVETO)
        start = points[0].copy()
        command = 'l'
    else:
        codes.append(Path.LINETO)

    current = points[-1]
    i += count

arrow = Path(vertices, codes)
```

```python
########################
##-Writing cycle plot-##
########################

fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])

ax.set_xlim(0, 1157)
ax.set_ylim(489, 0)
ax.set_aspect('equal')
ax.axis('off')

for i, (stage, (x, y)) in enumerate(zip(stages, centers)):

    ##-Position and rotate each arrow
    transform = Affine2D().translate(-373.225, -252.23)

    if i % 2 == 0:
        transform.scale(-1, 1).rotate_deg(82)
        number_y = y - 32
        name_y = y + 7
    else:
        transform.rotate_deg(70)
        number_y = y + 35
        name_y = y - 7

    transform.scale(2.20).translate(x, y)

    ax.add_patch(PathPatch(
        arrow,
        transform=transform + ax.transData,
        facecolor=to_rgba(green, 0.5),
        edgecolor=green,
        linewidth=1,
    ))

    ##-Stage number and name
    ax.text(x, number_y, str(i + 1),
            ha='center', va='center',
            fontsize=9, fontweight='bold', zorder=10)

    ax.text(x, name_y, stage,
            ha='center', va='center', fontsize=9, color='black', zorder=10)

##-Descriptions around the arrows
for x, y, alignment, label in notes:
    ax.text(x, y, label,
            ha=alignment, va='top',
            fontsize=9, color='black', linespacing=1.15, zorder=10)

##-Versioned exports
output_dir = FilePath('/content/inputs') if FilePath.cwd() == FilePath('/content') else FilePath.cwd() / 'inputs'

output_dir.mkdir(parents=True, exist_ok=True)
fig.savefig(output_dir / 'Figure_1_writing_cycle_v11.png', dpi=600, facecolor='white', transparent=False)
fig.savefig(output_dir / 'Figure_1_writing_cycle_v11.svg', facecolor='white', transparent=False)
fig.savefig(output_dir / 'Figure_1_writing_cycle_v11.pdf', facecolor='white', transparent=False)

plt.show()
```
